```python
import torch
```

```python

x=torch.tensor(3,requires_grad=True,dtype=torch.float32)
```

```python
y=x**2
```

```python
x
```

```python
y
```

```python
y.backward()
```

```python
x.grad
```

```python
x=torch.tensor(3,requires_grad=True,dtype=torch.float32)
y=x**2
z=torch.sin(y)
```

```python
x,y,z
```

```python
z.backward()
```

```python
z
```
